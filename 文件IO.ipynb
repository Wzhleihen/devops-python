```python
# 文件对象比较
# 每次打开都的到不一样的对象，地址不一样，但是内容是一样的，比较为false

import sys

f2 = open('test')
with f2 as f1:
    print(f1)
    print(f2)
    print(f1 == f2)
    print(f1 is f2)

print(f2.closed)
```

```python
f3 = open('test')
f4 = open('test')
# 两次打开得到两次文件描述符，文件描述符不一样，但是文件内容是一样的，比较为false

print(f3 == f4)  # 文件对象不可以直接比较内容，is是比较id
print(id(f3))
print(id(f4))

# 返回该文件对象在操作系统中的文件描述符编号
print(f3.fileno(), f4.fileno())


f3.close()
f4.close()
print(f3.closed)
print(f4.closed)
```

```python
# 除法符号的运算符重载
Path('a') / 'b' / 'c/d' 
```

```python
# Path对象 / Path对象
Path('a/b') / Path('c/d/e')
```

```python
# Path对象 / 字符串
Path('a/b') / 'c/d/e'
```

```python
# 字符串 / Path对象
'a/b' / Path('c/d/e')
```

```python
# 错误示范
# 'a/b' / 'c/d/e' / Path('f')
```

```python
#  Path 拼接用法：
# 1. Path对象 / Path对象
# 2. Path对象 / 字符串
# 3. 字符串 / Path对象
# 注意：除法符号运算符重载，只能用于拼接路径，不能用于运算，否则会报错

# 这里会使用Path的内部翻转机制，先计算 'c/d/e' / Path('f')，得到一个Path对象，然后再进行拼接
'a/b' / ('c/d/e' / Path('f'))

# 在工程代码中我会建议这样写：
Path('a/b') / 'c/d/e' / 'f'

```

```python

p1 = Path('a', 'b', 'c/b', Path('e', 'f/g')) 
# WindowsPath('a/b/c/b/e/f/g')  是视觉上的效果，实际是WindowsPath对象，内存中是指向一个对象的地址

p1   #这种表达方式称为 repr 
```

```python
str(p1)  # 'a\\b\\c\\b\\e\\f\\g'  windows系统下的路径字符串表现形式, 强制转换成字符串
```

```python
repr(p1)  # 以对象表现方式显示
str(p1)  # 以字符串表现方式显示
```

```python
# Path 的优化原理
open(p1)
# open函数的参数是字符串，所以这里会自动调用str(p1)

if isinstance(p1, Path):
    open(str(p1))
# isinstance(p1, Path) 判断p1是否是Path对象，如果是，则调用str(p1)方法，将Path对象转换成字符串，然后再传给open函数

# 随着 Python 生态系统的发展，越来越多的库开始支持 Path 对象作为文件路径的参数类型。
# 这种支持使得代码更加简洁和易读，同时也减少了手动转换路径字符串的需要。
# 因此，在实际开发中，建议优先使用 Path 对象来处理文件路径，而不是字符串形式的路径。
```

```python
from pathlib import Path

p = Path()
p = p / 'a'
p1 = 'b' / p
p2 = Path('c')
p3 = p2 / p1
print(p1, p2, p3)
print(p3.parts)  #  获取路径的各个部分，返回一个元组
print(p3.joinpath('d', 'e/f', Path('g/h')))  #  拼接路径，返回一个新的 Path 对象
```

```python
from pathlib import Path

p = Path('/etc/mysql/install/mysql.tar.gz')
print(p.parent)

for x in p.parents: # 可迭代对象
    print(x, type(x)) # 每一个parent 都是 Path 对象
    # print(x) == print(str(x)) 在面向高级对象编程中，更推荐使用对象本身，而不是其字符串表示形式

# p.parents[0]  # 获取上一级目录
# p.parents[1]  # 获取上上级目录
p.parents[-1]  # 在旧版本Python中不支持负索引，新版本负索引 从后往前数
```

```python
from pathlib import Path

# 注意： 扩展名，是可以改变的，在windows下，扩展名只是辅助识别文件类型的一个手段，在linux下，扩展名是文件类型的一部分.
p = Path('/etc/mysql/install/mysql.tar.gz')
print('1', p.parent) # 获取上一级目录
print('2', p.name)  # 获取文件名
# name = stem + suffix
print('3', p.stem)  # 获取不带扩展名的文件名
print('4', p.suffix)  # 获取文件扩展名
print('5', p.suffixes)  # 获取文件扩展名列表
print('6', p.with_name('my.cnf'))  #  返回一个新的 Path 对象，替换文件名
print('7', p.with_name('redis').with_suffix('.zip'))  #  返回一个新的 Path 对象，替换文件名和扩展名
```
