```python
class Person:
    def __init__(self, name):
        self.__name = name

    def get_name(self):
        return self.__name
    
    def set_name(self, new_name):
        self.__name = new_name

tom = Person("Tom")

# tom._name 不推荐直接访问私有属性
print(tom.get_name())  # 输出: Tom

# 修改名字
# tom._name = 'jerry' #不推荐
tom.set_name('jerry')
print(tom.get_name())  # 输出: jerry
```

# property 
是 Python 中的一个内置装饰器，用于将类的方法转换为属性访问的形式。通过使用 `@property` 装饰器，可以让一个方法像访问属性一样被调用，而不需要显式地调用方法。这种方式使得代码更加简洁和易读，同时也提供了更好的封装性。
使用 `@property` 装饰器的好处包括：
1. **简化访问**：可以像访问属性一样访问方法，代码更清晰。
2. **封装性**：可以在方法中添加逻辑来控制属性的
3. 访问和修改，增强了封装性。
4. **兼容性**：可以在不改变接口的情况下修改方法的实现，保持代码的兼容性。

getters 和 setters 是面向对象编程中的概念，分别用于获取和设置对象的属性值。通过使用 `@property` 和 `@<property_name>.setter` 装饰器，可以将方法定义为属性访问的形式，从而实现对属性的控制和封装。



```python
class Person:
    def __init__(self, name):
        self.__name = name
        
    @property
    def name(self):  # getter 方法
        return self.__name
    
    @name.setter
    def name(self, new_name):  # setter 方法
        self.__name = new_name

tom = Person("Tom")
print(tom.name)
tom.name = 'jerry'
print(tom.name)
```

```python
class Person:
    def __init__(self, name):
        self.__name = name
        

    def get_name(self):  # getter 方法
        return self.__name
    
    def set_name(self, new_name):  # setter 方法
        self.__name = new_name
    
    def del_name(self):  # 删除属性的方法
        del self.__name

    name = property(get_name, set_name, del_name, "name property documentation string.")

tom = Person("Tom")
print(tom.name)
tom.name = 'jerry'
print(tom.name)
```
