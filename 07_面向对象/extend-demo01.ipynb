```python
# 继承
# 继承是面向对象编程中的一个重要概念，
# 它允许我们创建一个新的类（子类）来继承一个现有的类（父类）的属性和方法。
# 通过继承，子类可以重用父类的代码，并且可以添加自己的属性和方法。
# 这样可以减少代码的重复，并且使得代码更加模块化和可维护。

class Animal:
    def shout(self):
        print("Animal is shouting")

class Cat(Animal):
    pass

a = Animal()
a.shout()
# 通过继承，Cat类自动获得了Animal类的shout方法，因此我们可以直接调用它。
c = Cat()
c.shout()
```

```python
class Animal:
    def __init__(self, name):
        self.__name = name
    
    def shout(self):
        print(f"{self.__name} shout's ")

class Cat(Animal): pass
class Dog(Animal): pass

a = Animal("animal")
a.shout()

c = Cat("cat")
c.shout()

d = Dog("dog")
d.shout()
```

```python
class Animal:
    __a = 10;  _b = 20;  c = 30;
    
    def __init__(self):
        self.__d = 40
        self._e = 50
        self.f = 60
        self.__a += 1
        
    def showa(self):
        print(self.__a)   # self._Animal__a
        print(self.__class__.__a)  #Cat._a => Cat._Animal_a 
    
    def __showb(self):
        print(self._b)
        print(self.__a)
        print(self.__class__.__a)

class Cat(Animal):
    __a = 100
    _b = 200


c = Cat()
c.showa()  # 11 , 10
#  解释：self.__a 使用父类的 __a 属性，而 self.__class__.__a 使用子类 Cat 的 __a 属性。

c._Animal__showb()  
print(c.c)
print(c._Animal__d)
print(c._e, c.f, c._Animal__a)
```

```python
print(c.__dict__)
print(c.__class__.__dict__.keys())
```

```python
class Animal:
    def shout(self):
        print("Animal is shouts")

class Cat(Animal):
    def shout(self):  # 重写父类的方法，覆盖
        # self.__class__.__base__.shout(self)  # 调用父类的方法
        # Animal.shout(self)  # 调用父类的方法
        super().shout()  # super() 函数返回一个临时对象，该对象允许我们调用最近父类的方法。
        print("miao")
        super(Cat, self).shout() # 在Python 2中，super() 函数需要两个参数：子类类名和子类实例。

# a = Animal()
# a.shout()

c = Cat()
c.shout()
```
