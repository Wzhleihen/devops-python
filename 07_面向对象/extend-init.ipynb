```python
class A:
    def __init__(self, a):
        self.a1 = a
        self.__d = 100  # 私有属性，无法被子类访问

# 当子类有自己的初始化方法时，父类的初始化方法不会被自动调用
# 子类实例创建属性虽然借用父类的构造器，但是创建当前实例是子类的实例，创建出来的对象仍然是“子类实例”。
class B(A):
    def __init__(self, b, c):
        super().__init__(b)  # 第一行显式调用父类的初始化方法
        self.b1 = b
        self.b2 = c
    
    def showme(self):
        print(self.b1, self.b2)
        print(self.a1)

c = B(1, 2)
c.showme()
```

```python
class A:
    def __init__(self, a):
        self.a1 = a
        self.__d = 100  # 私有属性，无法被子类访问

    def showme(self):
        print(self.a1, self.__d)
    
class B(A):
    def __init__(self, b, c):
        super().__init__(b)  # 显式调用父类的初始化方法
        self.b1 = b
        self.b2 = c
    
    def showme(self):
        super().showme()  # 调用父类的 showme 方法
        print(self.b1, self.b2)
        # print(self.__d) # 访问不到


c = B(1, 2)
c.showme()
```

```python
class A:
    def __init__(self):
        self.a1 = 100

class B(A):
    def __init__(self):
        super().__init__()
        self.b1 = 200

c = B()
print(c)
print(c.__dict__)
# 判断实例是否是某个类的实例，或者是否是某个类的子类的实例
print(isinstance(c, B),  isinstance(c, A)) 
```

```python
# 多态
# 在同一套接口下，不同的对象可以有不同的行为，这就是多态。
# 一般情况，多态的前提，继承和覆盖是必不可少的。
class Animal:
    def __init__(self, name):
        self.name = name

class Cat(Animal):
    def shout(self):
        print('miao')

class Dog(Animal):
    def shout(self):
        print('wangwang')

c = Cat('xiaohua')
c.shout()

d = Dog('ahuang')
d.shout()
```

```python
class  Document: # 抽象基类，在其他面向对象语言中，抽象基类不可以实例化, 抽象类一般不实例化，为了规范子类的行为，要求子类必须实现某些方法
    def __init__(self,content):
        self.content = content
    
    def print(self):  # 在python中相当于抽象方法，实际上没有实现，
        raise NotImplementedError('我是父类, 不实现打印方法, 子类必须实现 print 方法')

class Word(Document): pass

class Pdf(Document): pass

class PrintableWord(Word):
    def print(self):
        print('[ {self.content} ]')

class PrintablePdf(Pdf):
    def print(self):
        print('** {self.content} **')

# D = Document('hello world')
# D.print()

w = PrintableWord('hello world')
w.print()

p = PrintablePdf('hello world')
p.print()
```
