```python
# 常见的时间模块： datetime timezone timedelta time 
# datetime 模块提供了处理日期和时间的类和函数，可以进行日期和时间的计算、格式化和解析等操作。
# timezone 模块提供了处理时区的类和函数，可以获取当前时区、转换时区等操作。
# timedelta 模块提供了处理时间间隔的类和函数，可以计算两个时间之间的差值、增加或减少时间等操作。
# time 模块提供了处理时间的类和函数，可以获取当前时间、格式化和解析时间等操作。


import datetime  # datetime timezone timedelta time
```

```python
 #  class --> instance
 # 1 构造器
d1 = datetime.datetime(2024, 6, 1, 12, 0, 0)  # 创建一个 datetime 对象，表示 2024 年 6 月 1 日 12:00:00
```

```python
# 自定义时间
d1
```

```python
# 2 获取当前时间
d2 = datetime.datetime.now()  # 无时区时间，
```

```python
d2 
```

```python
# Unix 时间戳 timestamp 是指自 1970 年 1 月 1 日 00:00:00 UTC（协调世界时）以来的秒数。它是一种表示时间的方式，常用于计算机系统中。
# 获取当前时间的 Unix 时间戳
d3 = d2.timestamp()
```

```python
d3 
```

```python
d4 = datetime.datetime.fromtimestamp(d3)  # 从 Unix 时间戳创建 datetime 对象
```

```python
d4  
```

```python
# 在日常生活中，我们通常拿到的是字符串格式的时间，比如 "2024-06-01 12:00:00"。
# 我们可以使用 datetime 模块中的 strptime() 方法将字符串转换为 datetime 对象。
```

```python
datastring = "2024-06-01 12:00:00"

d5 = datetime.datetime.strptime(datastring, "%Y-%m-%d %H:%M:%S")  # 将字符串转换为 datetime 对象
```

```python
d5 ,type(d5)
```

```python
# 时间对象转换为字符串
d5.strftime("%Y-%m-%d %H:%M:%S")  # time --> format string
```

```python
d5  # naive datetime 对象，没有时区信息
```

```python
d6 = datetime.datetime(2024, 6, 1, 12, 0, 0, tzinfo=datetime.timezone.utc)  # 创建一个带有时区信息的 datetime 对象
```

```python
d6 # aware datetime 对象，包含时区信息
```

```python
# 构建一个偏差 8 小时的时区对象
datetime.datetime.now(datetime.timezone(datetime.timedelta(hours=8)))
```
