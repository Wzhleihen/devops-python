`````python
class Person:
    def __init__(self, name, age):  # 在函数方面可以用到五种参数，在方法方面也可以用到五种参数
        self.name = name   # 实例属性，实例变量，实例对象的属性，实例对象的变量
        self.age = age
    
    def showme(self):  # 类属性，方法，函数，方法Method，指向执行函数
        print(f'{self.name} is {self.age} years old')
''' 
实例属性和类属性的区别：
1. 实例属性，实例变量，实例对象的属性，实例对象的变量，是实例对象自己拥有的，实例对象之间属性值可以不同
2. 类属性，类变量，类对象的属性，类对象的变量，是类对象自己拥有的，实例对象之间属性值可以相同

有类不一定有实例，有实例一定有类
'''


# 实例化，指的是，类 构造 实例
tom = Person('tom', 20)
jerry = Person('jerry', 25)
`````

`````python
# 测试代码

class Person: 
    age = 3   
    height = 175
    
    def __init__(self, name, age=18):
        self.name = name
        self.age = age

tom = Person('Tom')
jerry = Person('Jerry', 20)

Person.age = 30
print(1, Person.age, tom.age, jerry.age) # 输出什么结果 

print(2, Person.height, tom.height, jerry.height) # 输出什么结果？ 
jerry.height = 180
print(3, Person.height, tom.height, jerry.height) # 输出什么结果？ 

tom.height += 10
print(4, Person.height, tom.height, jerry.height) # 输出什么结果？ 

Person.height += 15
print(5, Person.height, tom.height, jerry.height) # 输出什么结果？ 


Person.weight = 50
print(6, Person.weight, tom.weight, jerry.weight) # 输出什么结果？

print(7, tom.__dict__['height'])  # 可以查看吗？
# print(8, tom.__dict__['weight'])  # 可以查看吗？

# 实例通过属性访问规律：
# 实例.属性  优先访问实例自己的属性(实例字典)，，如果实例没有这个属性，就访问类的属性(类字典)，
# 如果类也没有这个属性，就继续访问父类的属性，直到找到为止，如果都没有找到，就抛出AttributeError异常

# tom.__dict__['height'] dict[key] KeyError: 'height'，实例对象的属性字典里没有这个属性，所以访问失败，抛出KeyError异常
# tom.height 访问成功，实例对象的属性字典里没有这个属性，所以继续
`````

`````python
# 3  类方法，@classmethod修饰的方法，类方法，单参函数
# 用的不多，通过实例或者访问方法时，方法内只需要类就可以的时候, 工具方法
# 类方法装饰器，会对实例或者类访问该方时，会绑定当前类
# 调用时，会自动注入第一参数为当前类
class A:
    @classmethod  # test = classmethod(test)
    def test(cls):
        print('test ~~~', type(cls), cls)

# 在java中，一般写clz，代表class,我们习惯写为cls, 不同语言习惯不同，python习惯写cls, 代表class
# self → 实例
# cls → 类

# 绑定的好处，在未来可以注入
`````

`````python
A.__dict__
`````

`````python
x = A()
`````

`````python
x.test()  # 方法绑定， 绑定类A
`````

`````python
A.test # 方法绑定， 绑定类A
`````

`````python
class B:
    @classmethod
    def foo(cls, x, y=20):
        print(cls, x, y)
`````

`````python
t = B()
`````

`````python
t.foo(10)
`````

`````python
B.foo(23, t)
`````

`````python
class C:
    @classmethod
    def foo(cls):
        print(cls)

t = C()
t.foo()
`````

`````python
# 4 静态方法，@staticmethod修饰的方法，静态方法，单参函数
# 用的不多，通过实例或者访问方法时，方法内不需要类也不需要
# 实例的时候，工具方法
# 静态方法装饰器，会对实例或者类访问该方时，会绑定当前类
# 调用时，不会自动注入参数
# 使用极少，我们编程基本不用，在源码，比较常见

class A:
    @staticmethod
    def test():
        print('test ~~~', type(A), A)

A.test()
A.__dict__
`````

`````python
class A:
    def __init__(self):
        self.value = 'abc'

    def showme(self):
        print(self)
        print(self.value)
        print(self.__class__)
        print(__class__)
        print(__class__.__name__)
        print(__name__)
`````

`````python
A.showme(100)
`````
